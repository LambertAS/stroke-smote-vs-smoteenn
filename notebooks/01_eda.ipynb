# Exploratory data analysis

Recreates the exploratory figures (Fig 2 to 4) with the functions in `src/stroke_imbalance/plots.py`. Run `python -m stroke_imbalance.data` first to download the dataset. The committed versions of these figures are in `results/paper/figures/`.

```python
from pathlib import Path

from IPython.display import Image, display

from stroke_imbalance.data import load_raw
from stroke_imbalance.plots import (
    plot_class_balance,
    plot_bmi_glucose_by_stroke,
    plot_numeric_distributions,
)

raw = load_raw()
out = Path('eda_figures')
raw.head()
```

## Missing values

Only BMI has missing values (201 of 5,110 rows).

```python
raw.isna().sum()
```

## Fig 2: class distribution

249 of 5,110 patients (4.87%) had a stroke.

```python
display(Image(plot_class_balance(raw, out / 'fig2_class_balance.png')))
```

## Fig 3: distribution of the numerical variables

```python
display(Image(plot_numeric_distributions(raw, out / 'fig3_numeric_distributions.png')))
```

## Fig 4: BMI and glucose by stroke status

Each class is normalized separately, so the curves compare shapes, not counts.

```python
display(Image(plot_bmi_glucose_by_stroke(raw, out / 'fig4_bmi_glucose_by_stroke.png')))
```
